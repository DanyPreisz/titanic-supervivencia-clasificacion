# Supervivencia en el Titanic — Clasificación

Proyecto de ciencia de datos listo para **Google Colab**.

**Pregunta de negocio:** ¿qué pasajeros tenían más probabilidad de sobrevivir y con qué variables se puede predecir?

**Target:** `Survived` (0 = no, 1 = sí)

**Stack:** pandas, numpy, matplotlib, seaborn, scikit-learn

Dataset público (mismo esquema que Kaggle Titanic):
`https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv`
